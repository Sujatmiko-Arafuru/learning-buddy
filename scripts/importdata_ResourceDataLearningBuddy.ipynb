```python
collection_names = [
    "Learning_Path_Answer",
    "Current_Interest_Questions",
    "Current_Tech_Questions",
    "Skill_Keywords",
    "Student_Progress"
]

collections = {}   # tempat menyimpan semua collection

for i, name in enumerate(collection_names, start=1):
    collections[i] = db.create_collection(name)
```
